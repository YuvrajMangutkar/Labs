***a. Implement Reinforcement Learning using an example of a maze environment that the agent needs to explore.***

```python
import numpy as np
import random
```

```python
rows=5
cols=5
start=(0,0)
goal=(4,4)

walls=[
    (0,3),
    (1,1),(1,3),
    (3,1),(3,3)

]
```

```python
actions={
    0:(-1,0),
    1:(1,0),
    2:(0,-1),
    3:(0,1)

}
q_table=np.zeros((25,4))
q_table
```

```python
def get_state(position):
    return position[0]*cols+position[1]
    
```

```python
def move(postion,action):
    dr,dc=actions[action]
    nr=position[0]+dr
    nc=position[1]+dc

    if nr<0 or nr>=rows or nc<0 or nc>=cols:
        return position,-5,False

    if (nr,nc) in walls:
        return position,-5,False
    
    if(nr,nc)==goal:
        return (nr,nc),100,True

    return (nr,nc),-1,False
    
```

```python
alpha = 0.1       # Learning rate
gamma = 0.9       # Discount factor
epsilon = 0.2     # Exploration probability
episodes = 5000

for episode in range(episodes):
    position = start

    for step in range(100):
        state = get_state(position)

        # Choose an action
        if random.random() < epsilon:
            action = random.randint(0, 3)
        else:
            action = np.argmax(q_table[state])

        # Perform the action
        next_position, reward, done = move(position, action)
        next_state = get_state(next_position)

        # Update Q-value
        q_table[state, action] = q_table[state, action] + alpha * (
            reward + gamma * np.max(q_table[next_state])
            - q_table[state, action]
        )

        position = next_position

        if done:
            break

print("Training completed!")
```

```python
position = start
path = [position]

for step in range(100):
    state = get_state(position)
    action = np.argmax(q_table[state])

    next_position, reward, done = move(position, action)

    if next_position == position:
        print("Agent got stuck!")
        break

    position = next_position
    path.append(position)

    if done:
        break

print("Learned path:", path)
print("Goal reached:", position == goal)
```
