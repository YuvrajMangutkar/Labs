```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#csv dataset
jan_data = {
    'Order_ID': [101, 102, 103, 104],
    'Customer_ID': [1, 2, 3, 1],
    'Product': ['Laptop', 'Mobile', 'Tablet', 'Laptop'],
    'Category': ['Electronics', 'Electronics',
                 'Electronics', 'Electronics'],
    'Quantity': [2, 1, 3, 1],
    'Price': [50000, 20000, 15000, 50000]
}

jan_df = pd.DataFrame(jan_data)
jan_df.to_csv('sales_jan.csv', index=False)
print("CSV file created successfully!")
```

```python
#Excel
feb_data = {
    'Order_ID': [105, 106, 107, 107],
    'Customer_ID': [4, 2, 5, 5],
    'Product': ['Headphones', 'Mobile', 'Tablet', 'Tablet'],
    'Category': ['Accessories', 'Electronics',
                 'Electronics', 'Electronics'],
    'Quantity': [2, 2, 1, 1],
    'Price': [2000, 20000, 15000, 15000]
}

feb_df = pd.DataFrame(feb_data)

feb_df.to_excel('sales_feb.xlsx', index=False)

print("Excel file created successfully!")
```

```python
#JSON 
mar_data = [
    {
        'Order_ID': 108,
        'Customer_ID': 3,
        'Product': 'Laptop',
        'Category': 'Electronics',
        'Quantity': 1,
        'Price': 50000
    },
    {
        'Order_ID': 109,
        'Customer_ID': 4,
        'Product': 'Headphones',
        'Category': 'Accessories',
        'Quantity': 3,
        'Price': 2000
    },
    {
        'Order_ID': 110,
        'Customer_ID': 5,
        'Product': 'Mobile',
        'Category': 'Electronics',
        'Quantity': None,
        'Price': 20000
    }
]

pd.DataFrame(mar_data).to_json(
    'sales_mar.json',
    orient='records',
    indent=4
)

print("JSON file created successfully!")
```

```python
df_csv = pd.read_csv('sales_jan.csv')

df_csv
```

```python
df_excel = pd.read_excel('sales_feb.xlsx')

df_excel
```

```python
df_json = pd.read_json('sales_mar.json')

df_json
```

```python
#Explore datasets
print("CSV DATA")
display(df_csv.head())

print("EXCEL DATA")
display(df_excel.head())

print("JSON DATA")
display(df_json.head())
```

```python
print("CSV Information")
df_csv.info()

print("\nExcel Information")
df_excel.info()

print("\nJSON Information")
df_json.info()
```

```python
print("CSV Missing Values:")
print(df_csv.isnull().sum())

print("\nExcel Missing Values:")
print(df_excel.isnull().sum())

print("\nJSON Missing Values:")
print(df_json.isnull().sum())
```

```python
print("CSV Duplicates:", df_csv.duplicated().sum())

print("Excel Duplicates:", df_excel.duplicated().sum())

print("JSON Duplicates:", df_json.duplicated().sum())
```

```python
df_csv = df_csv.drop_duplicates()

df_excel = df_excel.drop_duplicates()

df_json = df_json.drop_duplicates()

print("Duplicates removed successfully!")
```

```python
df_csv['Quantity'] = df_csv['Quantity'].fillna(
    df_csv['Quantity'].median()
)

df_excel['Quantity'] = df_excel['Quantity'].fillna(
    df_excel['Quantity'].median()
)

df_json['Quantity'] = df_json['Quantity'].fillna(
    df_json['Quantity'].median()
)

print("Missing values handled successfully!")
```

```python
#correcting data type
for data in [df_csv, df_excel, df_json]:
    data['Order_ID'] = data['Order_ID'].astype(int)
    data['Customer_ID'] = data['Customer_ID'].astype(int)
    data['Quantity'] = data['Quantity'].astype(int)
    data['Price'] = data['Price'].astype(float)

print("Data types corrected!")
```

```python
#verifying cleaned data
print("CSV Missing Values:")
print(df_csv.isnull().sum())

print("\nExcel Missing Values:")
print(df_excel.isnull().sum())

print("\nJSON Missing Values:")
print(df_json.isnull().sum())
```

```python
#combining datasets into one dataset
df = pd.concat(
    [df_csv, df_excel, df_json],
    ignore_index=True
)

df
```

```python
print("Final Dataset Shape:", df.shape)

print("\nFinal Dataset:")
display(df)

print("\nDataset Information:")
df.info()
```

```python
#data transformation
#merging
customer_data = {
    'Customer_ID': [1, 2, 3, 4, 5],
    'Customer_Name': ['Rahul', 'Amit', 'Priya',
                      'Sneha', 'Rohit'],
    'City': ['Pune', 'Mumbai', 'Kolhapur',
             'Nashik', 'Pune']
}

customers = pd.DataFrame(customer_data)

customers
df = pd.merge(
    df,
    customers,
    on='Customer_ID',
    how='inner'
)

df
```

```python
#derive new variable
df['Total_Sales'] = df['Quantity'] * df['Price']

df
```

```python
#spliting columns
df['First_Name'] = df['Customer_Name'].str.split().str[0]

df
```

```python
df['Sales_Status'] = np.where(
    df['Total_Sales'] >= 50000,
    'High Sales',
    'Low Sales'
)

df
```

```python
#sales data analysis
#descriptive statistics
df.describe()
```

```python
total_sales = df['Total_Sales'].sum()

print("Total Sales:", total_sales)
```

```python
average_order = df['Total_Sales'].mean()

print("Average Order Value:", average_order)
```

```python
total_quantity = df['Quantity'].sum()

print("Total Quantity Sold:", total_quantity)
```

```python
#productwise sales analysis

product_sales = df.groupby('Product')['Total_Sales'].sum()

print(product_sales)
```

```python
#citywise sales analysis
city_sales = df.groupby('City')['Total_Sales'].sum()

print(city_sales)
```

```python
#category wise sales distribution
category_sales = df.groupby('Category')['Total_Sales'].sum()

print(category_sales)
```

```python
#customerwise sales analysis
customer_sales = df.groupby('Customer_Name')['Total_Sales'].sum()

print(customer_sales)
```

```python
#Data Visualization
#Bar-plot(product wise sales)
plt.figure(figsize=(8, 5))

product_sales.plot(kind='bar', color='skyblue')

plt.title('Product-wise Total Sales')
plt.xlabel('Product')
plt.ylabel('Total Sales')

plt.xticks(rotation=45)
plt.show()
```

```python
#category wise sales 
plt.figure(figsize=(7, 7))

category_sales.plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Category-wise Sales Distribution')
plt.ylabel('')

plt.show()
```

```python
#box plot(sales distribution)
plt.figure(figsize=(8, 5))

sns.boxplot(x=df['Total_Sales'])

plt.title('Sales Distribution')
plt.xlabel('Total Sales')

plt.show()
```

```python
#customer wise sales distribution
plt.figure(figsize=(8, 5))

customer_sales.plot(kind='bar', color='orange')

plt.title('Customer-wise Sales')
plt.xlabel('Customer')
plt.ylabel('Total Sales')

plt.xticks(rotation=45)
plt.show()
```

```python
#store final cleaned analyzed dataset
df.to_csv('final_sales_data.csv', index=False)

print("Final cleaned dataset saved successfully!")
```
