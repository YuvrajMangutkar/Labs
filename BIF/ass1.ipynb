**1. Assignment: DNA Sequence Analysis. Task: Analyze a given DNA sequence and perform basic sequence
manipulation, including finding motifs, calculating GC content, and identifying coding regions. Deliverable: A
report summarizing the analysis results and any insights gained from the sequence.**

```python
#define the dna sequence
dna = "ATGCGATCGATGAAATAG"

print("DNA Sequence:", dna)
print("Sequence Length:", len(dna))
```

```python
#validate the dna sequence
dna = dna.upper()

if all(base in "ATGC" for base in dna):
    print("Valid DNA Sequence")
else:
    print("Invalid DNA Sequence")
```

```python
#lets go with specific dna pattern
motif = "ATG"

positions = []

for i in range(len(dna)):
    if dna[i:i+len(motif)] == motif:
        positions.append(i+1)

print("Motif:", motif)
print("Occurrences:", len(positions))
print("Positions:", positions)
```

```python
#calculate GC content
g = dna.count("G")
c = dna.count("C")

gc_content = ((g + c) / len(dna)) * 100

print("Guanine (G):", g)
print("Cytosine (C):", c)
print("GC Content:", round(gc_content, 2), "%")
```

```python
#identify coding regions(ORFs)
#open reading frames
start_codon = "ATG"
stop_codons = ["TAA", "TAG", "TGA"]

orfs = []

for frame in range(3):
    for i in range(frame, len(dna)-2, 3):

        if dna[i:i+3] == start_codon:

            for j in range(i+3, len(dna)-2, 3):

                if dna[j:j+3] in stop_codons:

                    orfs.append((i+1, j+3, dna[i:j+3]))
                    break

print("Possible Coding Regions:")

for orf in orfs:
    print("Start:", orf[0],
          "End:", orf[1],
          "Sequence:", orf[2])
```

```python
print("========== DNA ANALYSIS REPORT ==========")

print("DNA Sequence:", dna)
print("Sequence Length:", len(dna))

print("\nMotif Analysis")
print("Motif:", motif)
print("Occurrences:", len(positions))
print("Positions:", positions)

print("\nGC Content Analysis")
print("G Count:", g)
print("C Count:", c)
print("GC Content:", round(gc_content, 2), "%")

print("\nCoding Region Analysis")
print("Total ORFs Found:", len(orfs))

for number, orf in enumerate(orfs, 1):
    print(f"ORF {number}: {orf[2]}")
    print(f"Position: {orf[0]} - {orf[1]}")

print("\n========== ANALYSIS COMPLETED ==========")
```
