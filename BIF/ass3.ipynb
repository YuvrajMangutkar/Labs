**4. Assignment: Molecular Docking and Virtual Screening. Task: Perform molecular docking simulations to predict
the binding affinity between a protein target and a small molecule ligand. Additionally, conduct virtual screening
to identify potential drug candidates. Deliverable: A report summarizing the docking results, including the binding
poses and potential lead compounds.**

Our practical has four main tasks:
1. Select a target protein.
2. Select small molecule ligands (potential drugs).
3. Perform molecular docking to predict how ligands bind to the protein.
4. Perform virtual screening to identify the best potential drug candidates.

Molecular docking is a computational technique used to predict how a small molecule (ligand) interacts with a protein.

Binding affinity tells us how strongly a ligand is predicted to bind to a protein.

In AutoDock Vina, binding affinity is represented in kcal/mol.
- More negative score = More favorable predicted binding.
- Less negative score = Less favorable predicted binding.

Virtual screening means testing multiple ligand molecules computationally and ranking them based on their predicted binding scores

A binding pose describes the position and orientation of a ligand inside the protein's binding pocket

```python
import pandas as pd
```

```python
data = {
    "Ligand": ["Ligand_A", "Ligand_B", "Ligand_C",
               "Ligand_D", "Ligand_E", "Ligand_F"],

    "Hydrogen_Bonds": [3, 2, 4, 1, 3, 2],

    "Hydrophobic_Contacts": [5, 4, 6, 3, 7, 5],

    "Molecular_Weight": [310, 285, 340, 250, 360, 300],

    "Binding_Score": [-7.8, -6.4, -8.5, -5.2, -9.1, -7.0]
}
```

```python
ligands = pd.DataFrame(data)

print(ligands)
```

```python
ranked = ligands.sort_values("Binding_Score")

print(ranked.to_string(index=False))
```

```python
best = ranked.iloc[0]

print("Best candidate:", best["Ligand"])

print("Docking score:",
      best["Binding_Score"], "kcal/mol")
print("Docking scores are predictions and require experimental validation.")
```
