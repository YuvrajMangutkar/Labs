**Implement a program for retrieval of documents using inverted files.**

```python
import re
```

```python
documents={
    "D1":"Python is easy",
    "D2":"Python is powerful",
    "D3":"Data Science is easy"
}
documents
```

```python
inverted_index={}
for doc_id,text in documents.items():
    text=text.lower()
    words=re.findall(r'\b\w+\b',text)

    for word in words:
        if word not in inverted_index:
            inverted_index[word]=[]

        if doc_id not in inverted_index[word]:
            inverted_index[word].append(doc_id)
inverted_index
```

```python
query=input("Enter word to Search:").lower()
if query in inverted_index:
    for doc_id in inverted_index[query]:
        print(f"{doc_id}:{documents[doc_id]}")
else:
    print(f"No such Files or documents which contains the {query} term")
```
